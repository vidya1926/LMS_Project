```python
from sklearn.model_selection import train_test_split

# Use only labeled rows
labeled_df = df[df["label_overall"].notna()]

print("Total rows:", len(df))
print("Labeled rows:", len(labeled_df))

# Show a few labeled examples
labeled_df[["slide_index", "section_name", "label_overall", "obj_score", "concept_Score", "example_score", "load_score", "raw_text"]].head(10)


# Keep only rows that have a label
labeled_df = df[df["label_overall"].notna()].copy()

X = labeled_df["raw_text"]
y = labeled_df["label_overall"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

len(X_train), len(X_test)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("clf", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)
print("Model trained.")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)

print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification report:")
print(classification_report(y_test, y_pred))
```
